```python
import whisper

def transcribe_with_whisper(file_path, language="en"):
    # 加載模型，選擇 'base', 'small', 'medium', 'large' 等級別
    model = whisper.load_model("base")

    # 設定參數，將音訊轉文字
    result = model.transcribe(file_path, language=language)
    return result["text"]

# 英文語音檔案
chinese_text = transcribe_with_whisper("zh_male_01.wav", language="zh")
print("Chinese Text:", chinese_text) #label:感谢你，美丽的女士。琴，不坐下来喝一杯吗？我帮你也点了一份。
```

```python
import torch

# Get device
device = "cuda" if torch.cuda.is_available() else "cpu"


from TTS.api import TTS

# 初始化 TTS 模型
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(device)

tts.tts_to_file(text=chinese_text, speaker_wav="ko_male_05.wav", language="zh", file_path="ex4_from_ko_to_zh.wav")
```
