## Mongo 連線設置

```python
from pymongo import MongoClient

# 1. 連接到遠端 MongoDB
mongo_client = MongoClient("mongodb+srv://localhost:12027/")

# 2. 列出所有資料庫名稱
database_names = mongo_client.list_database_names()

```

```python
database_names
```

### Mongo 匯入資料

```python
import json

# 選擇資料庫
db = mongo_client["sample_traffic"]

# 選擇集合 (collection)
collection = db["myCollection"]

# 開啟 JSON 檔案並載入資料
with open('video_detect_output_example.json', 'r') as file:
    data = json.load(file)

# 如果 data 是一個列表，你可以一次匯入所有資料
if isinstance(data, list):
    collection.insert_many(data)
else:
    collection.insert_one(data)

print("資料匯入成功！")
```
