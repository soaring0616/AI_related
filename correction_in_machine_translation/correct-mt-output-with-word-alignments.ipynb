```python
import classla
classla.download('sl')
nlp = classla.Pipeline('sl', processors='tokenize,ner,pos,lemma')
```

```python
# Create a list of source texts
source_sentences = df_csv['text'].to_list()
len(source_sentences)
```

```python
# Create a list of translation sentences
target_sentences = df_csv['translation'].to_list()
```

```python
# Create a list of tokenized text where all words are separated by spaces
# this is needed for word alignment
tokenized_sentences = []

# Create a NER and lemma information into which you will save element for each sentence
proper_noun_list = []

# Run the processing on each sentence
for i in source_sentences:
    doc = nlp(i).to_dict()

    # Create a list of words in the current sentence
    current_sentence_list = []

    # Create a dictionary of all information on proper names for the current sentence
    word_dict = {}

    # Loop through the words in the sentence and add them to a current sentence list
    for sentence in doc:
        for word in sentence[0]:
            current_sentence_list.append(word["text"])

            # Find the current index - you need to substract 1 from the word index,
            # because indexing in the process starts with 1, not 0
            current_index = int(word["id"]) - 1

            # Add information on the lemma if the NE is personal name
            if "PER" in word["ner"]:
                word_dict[current_index] = [word["text"], word["lemma"]]

    # Join the list into a space-separated string to get tokenized text
    current_string = " ".join(current_sentence_list)

    tokenized_sentences.append(current_string)

    # Add the proper noun information for the current sentence to list for all sentences
    proper_noun_list.append(word_dict)

# Add the result to the df
df_csv["text_tokenized"] = tokenized_sentences
df_csv["proper_noun"] = proper_noun_list
```

```python
# Display the results
with pd.option_context('display.max_colwidth', None):
  display(df_csv[["text", "text_tokenized", "proper_noun", "translation"]][:2])
```

```python
import stanza

nlp = stanza.Pipeline(lang='en', processors='tokenize,mwt,pos,lemma, ner')
```

```python
#Create a list of tokenized text where all words are separated by spaces - this is needed for word alignment
tokenized_sentences = []

# Create a NER and lemma information into which you will save element for each sentence
proper_noun_list = []

# Run the processing on each sentence
for i in target_sentences:
    doc = nlp(i).to_dict()

    # Create a list of words in the current sentence
    current_sentence_list = []

    # Create a dictionary of all information on proper names for the current sentence
    word_dict = {}

    # Loop through the words in the sentence and add them to a current sentence list
    for sentence in doc:
        for word in sentence:
            current_sentence_list.append(word["text"])

            # Find the current index - you need to substract 1 from the word index,
            # because indexing in the process starts with 1, not 0
            if isinstance(word["id"], tuple):
              current_index = int(word["id"][0])
            else:
              current_index = int(word["id"]) - 1

            # Add information on the lemma if the NE is personal name
            if "ner" in word.keys() and "lemma" in word.keys() and "PER" in word["ner"]:
                word_dict[current_index] = [word["text"], word["lemma"]]

    # Join the list into a space-separated string to get tokenized text
    current_string = " ".join(current_sentence_list)

    tokenized_sentences.append(current_string)

    # Add the proper noun information for the current sentence to list for all sentences
    proper_noun_list.append(word_dict)

# Add the result to the df
df_csv["En_text_tokenized"] = tokenized_sentences
df_csv["En_proper_noun"] = proper_noun_list
```

## Apply word alignments with eflomal

```python
# Create files for all texts and all translations

Slovene_sentences = open("Slovene_sentences.txt", "w")
English_sentences = open("English_sentences.txt", "w")

for i in df_csv.text_tokenized.to_list():
	Slovene_sentences.write(i)
	Slovene_sentences.write("\n")

for i in df_csv.translation_tokenized.to_list():
	English_sentences.write(i)
	English_sentences.write("\n")

Slovene_sentences.close()
English_sentences.close()
```

```python
# Align sentences with eflomal
!python3 /content/eflomal/python/scripts/eflomal-align -s "/content/eflomal/Slovene_sentences.txt" -t "/content/eflomal/English_sentences.txt" --model 3 -f sl-en.fwd -r sl-en.rev
```

```python
# Get list of alignments from the eflomal's output
# Create a list with each line as a separate element
aligns_list = open("/content/eflomal/sl-en.rev", "r").readlines()

# Replace the \n element at the end of lines
aligns_list = [i.replace("\n", "") for i in aligns_list]

# Split each line into a list of alignment pairs: ["0-1", "1-2", "2-3"]
aligns_list = [i.split(" ") for i in aligns_list]

# Create a list to save the alignment dictionaries to
final_aligns = []

for i in aligns_list:
    # Create a dictionary for each sentence (each line)
    current_line = {}

    # For each alignment pair in line:
    for element in i:
        # Split the string "1-2" into a list of ["1", "2"]
        new_list = element.split("-")

        # Convert the strings to integers
        # and save a dictionary with the source word as key
        # and translated word as the value
        current_line[int(new_list[0])] = int(new_list[1])

    #Add the dictionary for each sentence to the list of all dictionaries
    final_aligns.append(current_line)

# Show result
print(final_aligns[:2])
# Output:
# [{0: 0, 1: 1, 2: 4, 3: 5, 4: 6, 6: 7, 7: 8, 9: 9, 10: 10, 11: 11}, {0: 2, 1: 3, 2: 4, 3: 5, 4: 6}]

# Add a to the df
df_csv["alignments"] = final_aligns
```

```python
# Substitute words in the translation based on alignments

# Create a list of tokenized translation, dictionary of proper nouns
# and alignments for each sentence pair
main_info = list(zip(df_csv["translation_tokenized"], df_csv["proper_noun"], df_csv["alignments"]))

# Create lists for the new translations and information on what was substituted
new_translations = []
substituted_only = []

# Add information whether an error occurred
error_list = []

# Loop through information for each sentence
for i in main_info:
    # Create a list of substitutions for current sentence
    current_substituted_only = []
    # Save error as No
    current_error = "No"

    # Take the relevant info for this sentence
    # from the main_info list
    current_translation = i[0]
    proper_noun_dict = i[1]
    alignments = i[2]

# If no proper names were detected, do not change the translation
    if proper_noun_dict == 0:
        new_translations.append(current_translation)

    else:
    # Else, substitute the word with the Slovene lemma based on the index - loop through the proper nouns in the dictionary to be changed
        for source_word_index in list(proper_noun_dict.keys()):
            try:
                # split the translation into list of words
                word_list = current_translation.split()

                # Get index of the translated word from the alignment dict
                translated_word_index = alignments[source_word_index]

                # Get the lemma to substitute the translated word with
                correct_lemma = proper_noun_dict[source_word_index][1]

                # If the translated word and lemma are not the same, get substituted word and its match
                if word_list[translated_word_index] != correct_lemma:

                    # Add the information that the word was substituted and with word it was substituted
                    current_substituted_only.append((word_list[translated_word_index], correct_lemma))

                  # Substitute the word in the translation word list
                 #  with the lemma based on its index
                    word_list[translated_word_index] = correct_lemma

                # After looping through proper names in the dict,
                # change the translation by merging the words back into a string
                current_translation = " ".join(word_list)

            # If the eflomal tool was not sure, it did not align a word
            # This can lead to errors, which we will save to the output
            except:
                current_error = f"Issue: index {word_index}: {i[1][word_index]}"

    # After the loop through proper nouns, save the new translation
    new_translations.append(current_translation)

  # Add information on what was substituted
    substituted_only.append(current_substituted_only)
    error_list.append(current_error)

# Add the lists to the df
df_csv["new_translations"] = new_translations
df_csv["substituted_words"] = substituted_only
df_csv["errors"] = error_list
```
