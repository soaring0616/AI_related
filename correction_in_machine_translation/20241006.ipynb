```python
update_ents = []

for idx in range(len(src_ents)):

  item = list(src_ents[idx])
  rm = list(remove[idx])
  ad = list(addition[idx])

  if len(rm) == 1 and rm[0]== '':
    pass
  else:
    for ii in rm:
      item.remove(ii)

  if len(ad) == 1 and ad[0]== '':
    pass
  else:
    for ii in ad:
      item.append(ii)

  update_ents.append(tuple(item))

print(update_ents)
```

```python
import jieba
import spacy
from spacy.tokens import Doc

nlp = spacy.load("zh_core_web_sm")
update_res = []

# 自定義分詞規則: 使用 jieba 添加自定義詞彙
def custom_tokenizer(text, custom_words, freq_words):
    for word in custom_words: # 添加自定義詞彙
      jieba.add_word(word)

    for word in freq_words: # 動態調整詞頻
      jieba.suggest_freq(word, tune=True)
    return list(jieba.lcut(text))

# 遍歷數據，應用自定義分詞規則
for idx in range(len(src_sent)):
    # 使用自定義分詞器
    tokens = custom_tokenizer(src_sent[idx], update_ents[idx], sugfreq_words[idx])
    # print("Jieba 分詞結果:", tokens)

    # 將分詞結果轉換為 spaCy 的 Doc 對象
    doc = Doc(nlp.vocab, words=tokens)

    # 輸出分詞結果
    update_res.append("/".join([token.text for token in doc]))


df = pd.DataFrame(list(zip(src_sent, src_sent_sep, remove, addition, update_ents, update_res)),
               columns =['原句', '原句的斷句', '移除', '需加入', '更新的分詞規則', '更新的分詞結果'])

df.head(30)
```
