## Stanza

```python
!pip install -q stanza
import stanza

nlp = stanza.Pipeline(lang='en', processors='tokenize,mwt,pos,lemma, ner')
```

```python
#Create a list of tokenized text where all words are separated by spaces - this is needed for word alignment
tokenized_sentences = []

# Create a NER and lemma information into which you will save element for each sentence
proper_noun_list = []

# Run the processing on each sentence
for i in target_sentences:
  doc = nlp(i).to_dict()

  # Create a list of words in the current sentence
  current_sentence_list = []

  # Create a dictionary of all information on proper names for the current sentence
  word_dict = {}


  # Loop through the words in the sentence and add them to a current sentence list
  for sentence in doc:
    for word in sentence:
      current_sentence_list.append(word["text"])

      # Find the current index - you need to substract 1 from the word index,
      # because indexing in the process starts with 1, not 0
      if isinstance(word["id"], tuple) and len(word["id"])>1:
        current_index = int(word["id"][0]) - 1
      else:
        current_index = int(word["id"]) - 1

      # Add information on the lemma if the NE is personal name
      if "ner" in word.keys() and "lemma" in word.keys() and "PER" in word["ner"]:
        word_dict[current_index] = [word["text"], word["lemma"]]

  # Join the list into a space-separated string to get tokenized text
  current_string = " ".join(current_sentence_list)

  tokenized_sentences.append(current_string)

  # Add the proper noun information for the current sentence to list for all sentences
  proper_noun_list.append(word_dict)

# Add the result to the df
results_df["En_text_tokenized"] = tokenized_sentences
results_df["En_proper_noun"] = proper_noun_list
```

```python
results_df.head()
```

```python
# Create a list of translation sentences
target_sentences = results_df['translation'].to_list()

# Create a list of tokenized text where all words are separated by spaces
# This is needed for word alignment
target_tokenized_sentences = []

# Run the language processing on each sentence
for i in target_sentences:
    # Transform the output into a dictionary
    doc = nlp(i).to_dict()

    # Create a list of words in the current sentence
    current_sentence_list = []

    # Loop through the words in the sentence
    # and add them to a current sentence list
    # to get tokenized sentences
    for sentence in doc:
        for word in sentence:
            # Add the word to the list of words for this sentence
            current_sentence_list.append(word["text"])

    # Join the list into a space-separated string to get tokenized text
    current_string = " ".join(current_sentence_list)

    # Add the tokenized sentence to a list of all sentences
    target_tokenized_sentences.append(current_string)

# Add the result to the df
results_df["translation_tokenized"] = target_tokenized_sentences
```
