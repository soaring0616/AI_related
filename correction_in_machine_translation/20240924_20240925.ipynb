```python
update_ents = []

for idx in range(len(src_ents)):

  item = list(src_ents[idx])
  rm = list(remove[idx])
  ad = list(addition[idx])

  if len(rm) == 1 and rm[0]== '':
    pass
  else:
    for ii in rm:
      item.remove(ii)

  if len(ad) == 1 and ad[0]== '':
    pass
  else:
    for ii in ad:
      item.append(ii)

  update_ents.append(tuple(item))

print(update_ents)
```

```python
import spacy
from spacy.tokenizer import Tokenizer
from spacy.lang.zh import Chinese
from spacy.symbols import ORTH

# Example sentence
# sentence = "Facebook的隱私政策再次引發爭議。"

# Create a Chinese nlp object
nlp = Chinese()

# Define custom tokenization rules
# rule = {'Facebook': ['Facebook'], '隱私政策': ['隱私政策']}

# Define a custom tokenizer
def custom_tokenizer(nlp):
    # Create the default tokenizer
    tokenizer = Tokenizer(nlp.vocab)

    # Add special cases (e.g., 'Facebook' and '隱私政策' as single tokens)
    for token in rule:
        tokenizer.add_special_case(token, [{ORTH: token}])

    return tokenizer

# Set the custom tokenizer
nlp.tokenizer = custom_tokenizer(nlp)

# Process the sentence
doc = nlp(sentence)

# Print the tokens
for token in doc:
    print(token.text)
```

```python
#(馬雲, 阿里巴巴, 中國)
# 馬雲創辦的阿里巴巴是中國最大的電商平台之一。

import spacy

# 加載 spaCy 的中文模型
nlp = spacy.load("zh_core_web_sm")

# 添加 entity_ruler 管道
ruler = nlp.add_pipe("entity_ruler", before="ner")
patterns = [{"label": "ORG", "pattern": "Facebook"},
            {"label": "LAW", "pattern": "隱私政策"}]

ruler.add_patterns(patterns)

# 定義文本
doc = nlp("Facebook的隱私政策再次引發爭議。")

# 遍歷 token，將實體作為一個單位，非實體逐詞分割
tokens = []
i = 0
while i < len(doc):
    if doc[i].ent_iob_ == 'B':  # 如果是實體的開始
        entity = []
        # 將整個實體加入 tokens 列表
        while i < len(doc) and doc[i].ent_iob_ != 'O':  # 直到實體結束
            entity.append(doc[i].text)
            i += 1
        tokens.append("".join(entity))  # 合併實體的所有部分
    else:
        tokens.append(doc[i].text)
        i += 1

# 按行打印每個 token 或實體
for token in tokens:
    print(token)
```
