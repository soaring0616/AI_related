```python
import whisper

model = whisper.load_model("turbo")
result = model.transcribe("0210.m4a")
#print(result["text"])
```

```python
groq_api_key = "you-groq-api-key"

from langchain_groq import ChatGroq

chat = ChatGroq(
    api_key = groq_api_key,
    model_name = "deepseek-r1-distill-llama-70b" # mixtral-8x7b-32768 字數會超過
)

from langchain_core.prompts import ChatPromptTemplate

system = "You are a helpful assistant."

# 分段總結 
# human = "The following text is the transcription script from the bible-reading meeting of Matthew 24:1-14 by AI model whisper,\
#         Please give me a series of segmented summary in Traditional Chinese: {text}"

# 訊息回應
human = "The following text is the transcription script from the bible-reading meeting of Matthew 24:1-14 by AI model whisper,\
        Please give me a promoting post of social media  in Traditional Chinese: {text}"

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", system), ("human", human)
    ]
)
```

```python
from langchain_core.output_parsers import StrOutputParser
chain = prompt | chat | StrOutputParser()


chain.invoke(
    {"text":result["text"]}
)
```
