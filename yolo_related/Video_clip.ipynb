```python
# loading video
clip1 = VideoFileClip("drive/My Drive/part1.mp4")
clip3 = VideoFileClip("drive/My Drive/part3.mp4")
clip4 = VideoFileClip("drive/My Drive/part4.mp4")

# concatenating both the clips
final = concatenate_videoclips([clip1,clip3,clip4])
```

```python
# saving the clip
final.write_videofile("drive/My Drive/final.mp4")
```

```python
# Import everything needed to edit video clips
from moviepy.editor import *

# loading video dsa gfg intro video
clip = VideoFileClip("drive/My Drive/final_new.mp4")

# loading audio file
audioclip = AudioFileClip("drive/My Drive/new.mp3")

# adding audio to the video clip
videoclip = clip.set_audio(audioclip)

videoclip.write_videofile("drive/My Drive/final_new1.mp4")
```

```python
# Importing the PIL library
from PIL import Image,ImageDraw,ImageFont

img = Image.open("drive/My Drive/output_folder/Sequence 01.00_00_04_16.Still113.jpg")

#Relative Path
#Image which we want to paste
img2 = Image.open("drive/My Drive/test_photo.jpg")
newsize = (242, 300)
img_resized = img2.resize(newsize)
img.paste(img_resized, (110, 90))

# Call draw Method to add 2D graphis in an image
I1 = ImageDraw.Draw(img)

# Custom font style and font size
myFont = ImageFont.truetype("drive/My Drive/HELVETICAINSERATLTSTD-ROMAN.OTF", 60)

# Add Text to an image
I1.text((480, 230), "KUNAL\nBHARADKAR", font=myFont, fill =(245, 169, 6),align ="center")

# Save the edited image
img.save("drive/My Drive/car2.jpg")
img2 = Image.open("drive/My Drive/car2.jpg")
img2
```
